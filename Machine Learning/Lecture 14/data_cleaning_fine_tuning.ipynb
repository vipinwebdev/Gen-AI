### Data Cleaning in ML, AI Data Management, & Fine Tuning

- The transition from a Business Problem to an AI Solution is entirely dependent on the quality of the data gathered.
- Key Concept: GIGO (Garbage In, Garbage Out).

- RAG Application:
   - Since RAG relies on "retrieving" facts, if the database contains "garbage" (errors), the LLM will generate confident but incorrect answers.

1. The 6 Pillars of Data Quality
- From the whiteboard, we've identified the critical issues you must "clean" before feeding data into an AI model or a Vector Database:

| Pillar       | Definition                         | Whiteboard                                      | Example                                                                 |
|--------------|------------------------------------|--------------------------------------------------|-------------------------------------------------------------------------|
| Validity     | Adherence to rules/formats         | Salary must be an integer; Height must be positive | Salary = "abc" ❌, Height = -170 ❌                                      |
| Accuracy     | Correctness of the information     | A height of 173 cm is valid, but is it correct?  | Stored value = 173 cm, actual value = 180 cm ❌                          |
| Completeness | No missing information             | Handling null or placeholder values              | Salary = NULL or 9999 ❌                                                |
| Duplicates   | No repeated records                | Removing duplicate entries                       | "Ankit" appears twice in dataset ❌                                      |
| Outliers     | Extreme values that skew results   | Identifying abnormal values                      | Ranbir earning 2 Crore vs others earning 7–8 LPA ⚠️                      |
| Data-Type    | Correct technical format          | Ensuring proper data types                       | Age stored as string "25" instead of int32 ❌                            |

### Fine-Tuning vs Basic AI Model (Knowledge Injection)

1. 🩺 The "Doctor" Analogy

- Basic AI Model → Like a student with general education
- Training with Domain Data → Like studying medicine
- Fine-Tuned Model → Like a specialized doctor

- `Basic AI Model + Medical Data → Training → Fine-Tuned AI Model (Doctor Bot)`
- Fine-tuning transforms a general-purpose model into a domain expert.
---
2. ⚖️ RAG vs Fine-Tuning (Decision Framework)

| Method                                   | Concept (Whiteboard)                                           | When to Use                                   |
| ---------------------------------------- | -------------------------------------------------------------- | --------------------------------------------- |
| **Fine-Tuning**                          | Train model using domain-specific dataset (e.g., medical file) | When behavior, tone, or expertise must change |
| **RAG** (Retrieval-Augmented Generation) | Provide external knowledge base without retraining             | When data must stay **updated & dynamic**     |

##### Simple Understanding:
- Fine-Tuning = Learn permanently
- RAG = Look up when needed
