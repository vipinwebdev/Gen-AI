### RAG (Retrieval-Augmented Generation)

- In simple terms, it's a way to give an AI model specific, reliable information (like your own documents) to look at before it answers your questions.
- This keeps the AI **"grounded"** and prevents it from making things up!

---

#### 🔍 How RAG Works (Step-by-Step)

1. **User asks a question**
   Example: *“What is our company’s leave policy?”*

2. **Retriever searches data sources**
   - Documents (PDFs, FAQs, databases, APIs)
   - Finds the most relevant chunks of information

3. **Relevant data is fetched**
   - Only useful context is selected (not the whole database)

4. **LLM generates answer using that data**
   - Combines retrieved info + language understanding
   - Produces a **context-aware and accurate response**

---

#### 🧠 Key Components of RAG

- **Retriever**
  - Searches and fetches relevant data
  - Uses embeddings + vector databases

- **Knowledge Base**
  - Your custom data (documents, APIs, DBs)

- **Generator (LLM)**
  - Generates final answer using retrieved context

---

#### 💡 Real Example

👉 **Company Chatbot**

**Without RAG:**
- AI may guess or hallucinate policies

**With RAG:**
- Fetches exact policy document
- Answers accurately:
  *“Employees are entitled to 18 paid leaves annually…”*

---

#### ⚙️ Simple RAG Flow (Technical)
User Query → Convert to Embedding → Search Vector DB → Retrieve Relevant Docs → Pass to LLM → Generate Answer



### 💻 Example Code (Basic Concept)

```python
from sentence_transformers import SentenceTransformer
import numpy as np

# Sample documents
docs = [
    "Employees get 18 paid leaves per year.",
    "Office timings are 9 AM to 6 PM.",
    "Work from home is allowed twice a week."
]

# Load model
model = SentenceTransformer('all-MiniLM-L6-v2')

# Convert docs to embeddings
doc_embeddings = model.encode(docs)

# User query
query = "How many leaves do employees get?"
query_embedding = model.encode([query])

# Find most similar doc
similarity = np.dot(doc_embeddings, query_embedding.T)
best_match = docs[np.argmax(similarity)]

print("Retrieved Context:", best_match)
