## Types Of Machine Learning

 ### 1. Supervised Learning (Learning from labeled examples)

- Supervised Learning is a type of Machine Learning in which the model is trained using a dataset that contains both the input data and the correct output labels.
- The goal of the model is to learn the relationship between input and outputs so that it can accurately predict the output for new unseen data.
- In this approach, the model continuously compares its predictions with the actual answers and adjust itself to minimize errors. Over time, it becomes better at making accurate predictions.

---
→ Example:
   - Suppose we train a model using past data of house prices:
      - input : Size, location, number of rooms.
      - output : price.
   - The model learns patterns like:
        - “Bigger house in a good location → higher price”
        -   Later, it can predict the price of a new house.
---
- Where it fits well:
    - Spam detection
    - Fraud detection
    - Medical diagnosis
    - Price prediction
---
Types inside Supervised Learning:
1. Regression (Predict continuous value)
    - Predict numbers
- 👉 Examples:
- House price prediction
- Salary prediction
2. Classification (Predict category)
    - Predict labels/classes
- 👉 Examples:
    - Spam detection
    - Disease prediction (Yes/No)

---
 - Algorithms:
    - Linear Regression
    - Logistic Regression
    - Decision Trees
    - Random Forest
    - Support Vector Machine (SVM)

### 2. Unsupervised Learning (Discovering hidden patterns)

- Unsupervised Learning is a type of machine learning where the model is given only input data without ant labeled outputs.
- The model's objective is to analyze the data and identify underlying structures, patterns, or grouping on its own.
- Since there are no predefined answers, the model explores similarities and differences within the data to organize it meaningfully.
---
→ Example
   - We have customer data but no labels.
      - Age , income, purchase behavior.
   - The model groups customers into clusters like:
      - High spenders
      - Budget Buyers
      - Occasional
  - This helps businesses understand customer segments without prior labeling.
---

- Where it fits well:
   - Customer segmentation
   - Recommendation systems
   - Pattern detection
   - Market basket analysis
---
Types inside Unsupervised:
   1. Clustering
   - Group similar data

- 👉 Example:
    - Customer segmentation (K-Means)
2. Association
    - Find relationships

- 👉 Example:
    - Market Basket Analysis
        (“People who buy bread also buy butter”)
---
- Algorithms:
    - K-Means
    - Hierarchical Clustering
    - Apriori Algorithm



### 3. Semi-Supervised Learning (Learning from limited labels)

- Semi-Supervised Learning is a hybrid approach where the model is trained using a small amount of labeled data combined with a large amount of unlabeled data.
- It leverages the labeled data to understand basic patterns and then uses the unlabeled data to improve its learning and generalization.
- This approach is useful when labeling data is expensive, time-consuming, or requires expert knowledge.
---
- Example:
    - In image recognition:
        - 500 images are labeled (cat/dog)
        - 50,000 images are unlabeled
    - The model uses the labeled data to learn initial features and then refines its understanding using the unlabeled images.
---
- Where it fits well:
    - Medical imaging.
    - Speech recognition.
    - Large-scale image classification.

### 4. Reinforcement Learning (Learning through interaction and rewards)
- Reinforcement Learning is a type of machine learning where an agent learns by interacting with an environment and receiving rewards or penalties based on its actions.
- The objective is to learn a strategy (policy) that maximizes the total reward over time.
- Instead of being told the correct answer, the model learns from experience by trying different actions and observing their outcomes.
- --
- Example:
    - In a game:
        - Correct move → reward
        - Wrong move → penalty

    - Over time, the model learns the best sequence of moves to win the game.

- Real-world applications:
    - Robotics
    - Self-driving cars
    - Game AI
    - Recommendation engines
