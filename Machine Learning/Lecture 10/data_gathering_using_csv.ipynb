## Data Gathering Using CSV

**`pandas.read_csv()`** is the main function used to load CSV files into a pandas DataFrame.

This notebook is a study guide for:

- the main use cases of `read_csv()`
- the full parameter list available in the pandas version used here
- practical examples using files from the `ML_Datasets` folder

---

### Core idea

- **Input:** a CSV file, URL, or file-like object
- **Output:** a pandas DataFrame
- **Goal:** control how raw CSV text becomes usable tabular data


### Common use cases of `read_csv()`

- **Load standard comma-separated files**
- **Read files with another separator** such as `;`, `|`, `\t`, or a whitespace pattern
- **Read only selected columns** with `usecols`
- **Assign custom column names** with `names`
- **Use a column as the index** with `index_col`
- **Parse date columns** into datetime values
- **Clean currency or formatted text while reading** with `converters`
- **Handle missing values** with `na_values` and `keep_default_na`
- **Force correct data types** with `dtype`
- **Read partial data** with `nrows`
- **Read large files in chunks** with `chunksize`
- **Handle messy files** with `comment`, `skiprows`, `skipfooter`, `on_bad_lines`, and encoding options
- **Read compressed or remote files** using `compression` and remote storage options

---

### Basic syntax

```python
import pandas as pd

df = pd.read_csv("path/to/file.csv")
```

---

### Example datasets used in this notebook

- `Chocolate_Sales.csv`
- `Global-YouTube-Statistics.csv`

````python
import pandas as pd

# Example 1: CSV with a BOM in the first header and a formatted currency column
sales = pd.read_csv(
    "../ML_Datasets/Chocolate_Sales.csv",
    encoding="utf-8-sig",
    converters={"Amount": lambda x: float(x.replace("$", "").replace(",", "").strip())}
)

sales.head()
````

### Full parameter guide for `read_csv()`

Below is the complete parameter list for the pandas version available in this workspace. The parameters are grouped by what they control so the list is easier to study.

---

### 1) File source and column structure

- **`filepath_or_buffer`**: The CSV file path, URL, or file-like object to read.
- **`sep`**: The separator between fields. Use this for comma, tab, pipe, semicolon, or other delimiters.
- **`delimiter`**: Same purpose as `sep`. Use one of them, not both.
- **`header`**: Which row should be used as the header row. `0` means the first row. `None` means the file has no header.
- **`names`**: Custom column names to assign manually.
- **`index_col`**: Column(s) to use as the DataFrame index.
- **`usecols`**: Read only a subset of columns. Useful for performance and memory savings.
- **`dtype`**: Force specific column data types.
- **`engine`**: Parsing engine to use. Usually pandas infers this automatically.
- **`converters`**: Functions applied to specific columns while reading.

---

### 2) Row selection and file size control

- **`skiprows`**: Skip one or more rows at the start, or skip rows based on a function.
- **`skipfooter`**: Skip rows from the end of the file.
- **`nrows`**: Read only the first `n` rows.
- **`chunksize`**: Read the file in chunks instead of loading everything at once.
- **`iterator`**: Return an iterator object for manual chunk processing.

---

### 3) Missing values and blank lines

- **`na_values`**: Extra values that should be treated as missing.
- **`keep_default_na`**: Keep pandas default missing-value markers. Set this to `False` when you want full control.
- **`na_filter`**: Turn missing-value detection on or off.
- **`skip_blank_lines`**: Ignore blank lines in the file.
- **`verbose`**: Print extra information about missing-value parsing.

---

### 4) Delimiters, spacing, and parsing behavior

- **`skipinitialspace`**: Ignore spaces that appear after the delimiter.
- **`lineterminator`**: Custom line-ending character.
- **`comment`**: Treat everything after this character as a comment.
- **`delim_whitespace`**: Use whitespace as the delimiter. Helpful for space-separated text files.
- **`dialect`**: Use a CSV dialect configuration.

---

### 5) Date and time parsing

- **`parse_dates`**: Parse one or more columns as dates.
- **`infer_datetime_format`**: Ask pandas to infer datetime formats. This is mainly a compatibility option in newer versions.
- **`keep_date_col`**: Keep the original columns when date columns are combined.
- **`date_parser`**: Custom function for parsing dates. Usually less preferred than pandas inference or `date_format`.
- **`date_format`**: Explicit date format string or mapping for parsing dates.
- **`dayfirst`**: Interpret dates like `01/02/2024` as day-first.
- **`cache_dates`**: Cache repeated date values for speed.

---

### 6) Boolean and value mapping

- **`true_values`**: Strings to treat as `True`.
- **`false_values`**: Strings to treat as `False`.

---

### 7) Text and numeric formatting

- **`thousands`**: Character used as the thousands separator.
- **`decimal`**: Character used as the decimal marker.
- **`quotechar`**: Character used to quote fields.
- **`quoting`**: CSV quoting mode.
- **`doublequote`**: Whether doubled quote characters inside a field represent one literal quote.
- **`escapechar`**: Escape character for special symbols.
- **`float_precision`**: Float parsing precision option.

---

### 8) Encoding and storage

- **`encoding`**: File encoding such as `utf-8`, `utf-8-sig`, `latin1`, or `cp1252`.
- **`encoding_errors`**: How decoding errors should be handled.
- **`compression`**: Compression format such as `infer`, `gzip`, `zip`, `bz2`, `xz`, or `zstd`.
- **`storage_options`**: Extra options for remote or cloud storage backends.
- **`memory_map`**: Use memory-mapped file access when possible.

---

### 9) Bad line and error handling

- **`on_bad_lines`**: What to do when a row has the wrong number of fields. Typical values are `error`, `warn`, and `skip`.

---

### 10) Performance and memory behavior

- **`low_memory`**: Process the file in smaller internal chunks to reduce memory pressure.
- **`dtype_backend`**: Choose the dtype backend, including nullable dtypes in newer pandas versions.


````python
# Example 2: large real-world CSV, limited columns and explicit dtypes
youtube = pd.read_csv(
    "../ML_Datasets/Global-YouTube-Statistics.csv",
    usecols=["rank", "Youtuber", "subscribers", "category", "Country"],
    dtype={"rank": "int64", "subscribers": "int64"},
    nrows=1000
)

youtube.head()
````

````python
# Example 3: missing-value handling on a data-cleaning workload
students = pd.read_csv(
    "../ML_Datasets/student_depression_dataset.csv",
    na_values=["", "NA", "N/A", "null"],
    keep_default_na=True
)

students.head()
````

### Practical parameter combinations

- **Load only the columns you need** when exploring a large file.
- **Use `nrows` first** when you want a fast preview.
- **Use `parse_dates` and `converters`** when raw values need cleanup during import.
- **Use `na_values`** when placeholder strings such as `NA`, `N/A`, or empty strings appear in the source.
- **Use `chunksize`** when the dataset is too large to load in one shot.
- **Use `encoding='utf-8-sig'`** when the first header contains odd characters from spreadsheet exports.
- **Use `sep` or `delimiter`** first when the file is not comma-separated.

---

### Notes on reading behavior

- `read_csv()` is flexible enough to handle both clean and messy CSV files.
- The fastest way to improve a failing import is usually to check the separator, encoding, header row, and missing-value markers.
- The most useful parameters for day-to-day work are often `usecols`, `dtype`, `parse_dates`, `na_values`, `nrows`, and `chunksize`.


````python
# Example 4: chunked reading for large files
reader = pd.read_csv("../ML_Datasets/Global-YouTube-Statistics.csv", chunksize=5000)
first_chunk = next(reader)
first_chunk.shape
````
