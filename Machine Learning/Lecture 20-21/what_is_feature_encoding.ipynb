# Feature Encoding in Machine Learning

##  What is Feature Encoding?

Feature Encoding is the process of converting **categorical (text-based) data into numerical format** so that it can be understood by machine learning models.

In simple terms:
> Machines understand numbers, not text.
> So we convert **text data → numeric data**.

---

##  Why Feature Encoding is Needed?

Machine Learning algorithms (like Linear Regression, Decision Trees, Neural Networks) work with **mathematical computations**.

They **cannot process raw text values** like:
- "Male", "Female"
- "Red", "Blue"
- "High", "Medium", "Low"

So we need to convert them into numbers.

---

###  Reasons for Feature Encoding:

1. **Algorithms require numerical input**
   - ML models perform calculations (distance, gradients, etc.)
   - Text cannot be used directly in formulas

2. **Maintain data consistency**
   - Standard numeric format helps models learn patterns better

3. **Avoid misinterpretation**
   - Proper encoding ensures relationships are represented correctly

---

##  Types of Data in ML

| Type              | Example                  | Needs Encoding? |
|------------------|--------------------------|---------------|
| Numerical        | Age, Salary              |  No          |
| Categorical      | Gender, Color, City      |  Yes         |



---

##  Types of Feature Encoding

#### 1. Label Encoding

- Converts categories into numbers
- Assigns each unique value a number

#### Example:
| Gender | Encoded |
|--------|--------|
| Male   | 0      |
| Female | 1      |

#### Python:
```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
df['Gender'] = le.fit_transform(df['Gender'])
```
---
#### 2. One-Hot Encoding
- Creates separate binary columns for each category
- One-Hot Encoding is a technique used to convert **categorical (text) data into numerical format** by creating **separate binary (0/1) columns** for each category.

- Each category becomes a **new column**, and:
    - **1** → category is present
    - **0** → category is not present

| Color | Red | Blue | Green |
| ----- | --- | ---- | ----- |
| Red   | 1   | 0    | 0     |
| Blue  | 0   | 1    | 0     |


##  Example
### Original Data:
| Color |
|-------|
| Red   |
| Blue  |
| Green |

### After One-Hot Encoding:

| Color_Red | Color_Blue | Color_Green |
|-----------|-----------|-------------|
| 1         | 0         | 0           |
| 0         | 1         | 0           |
| 0         | 0         | 1           |

#### How It Works

- For each unique category:
1. Create a new column
2. Mark **1 for matching category**, else **0**

---

#### 3. Ordinal Encoding
- Used when categories have natural order
- Example:
| Size   | Encoded |
| ------ | ------- |
| Small  | 1       |
| Medium | 2       |
| Large  | 3       |
```python
from sklearn.preprocessing import OrdinalEncoder

oe = OrdinalEncoder()
df[['Size']] = oe.fit_transform(df[['Size']])
```

---
#### 4. Frequency / Count Encoding
- Replace category with its frequency
- Example:
| City   | Frequency |
| ------ | --------- |
| Delhi  | 50        |
| Mumbai | 30        |

---
#### 5. Target Encoding (Advanced)
- Replace category with mean of target variable.
- Real-Life Example:
    - Dataset:
        - | Name | Gender | Purchased |
          | ---- | ------ | --------- |
          | A    | Male   | Yes       |
          | B    | Female | No        |
    - After Encoding:
        - | Gender | Purchased |
          | ------ | --------- |
          | 0      | 1         |
          | 1      | 0         |
