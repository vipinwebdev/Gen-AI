```python
import  pandas as pd
```

```python
bank = pd.read_csv("../ML_Datasets/Bank_Marketing_Original_Data.csv")
```

```python
bank.sample(3)
```

- Ordinal Column ---> Education
- Label Column ---> TARGET
- Nominal Column ---> Married, ....

```python
bank['Job'].value_counts()
```

```python
ordinal_data = bank[['Education']]
```

```python
target_data = bank[['TARGET']]
```

```python
from sklearn.preprocessing import LabelEncoder, OrdinalEncoder
le = LabelEncoder()
oe = OrdinalEncoder(categories=[['unknown', 'primary', 'secondary', 'tertiary']])
```

```python
ordinal_data = oe.fit_transform(ordinal_data)
```

```python
ordinal_data
```

```python
target_data = le.fit_transform(target_data)
```

```python
target_data
```
