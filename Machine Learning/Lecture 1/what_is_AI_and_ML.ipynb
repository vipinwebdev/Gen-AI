## AI & ML

### What is Artificial Intelligence ?

-   Artificial intelligence is the ability od machines (computers/software) to perform task normally require human intelligence.
-   That includes:
    - Thinking / Decision-Making.
    - Learning from experience.
    - Understanding languages.
    - Recognizing Patterns.

`AI = Machines that can learn, think and act like humans (to some extent).`

- AI is not about making machines “alive.”
- It’s about making machines use data to make smart decisions.
- --
#### AI → ML → Deep Learning → GenAI (relationship)
   - Artificial Intelligence (big umbrella)
        - Machine Learning (ML) → learns from data
            - Deep Learning → neural networks (like brain structure)
                - GenAI

### What is Machine Learning?
- Machine Learning (ML) is a subset of Artificial Intelligence where systems `learn from data and improve their performance without being explicitly programmed for every rule.`
- `ML = Instead of writing rules, we let the system learn rules from data`.
---
#### ⇒ Data
- Data = Experience
- Data is what the model learns from:
    - Without data → noting to learn
    - More quality data → better learning

- Example:
    - Emails → spam detection
    - Images → Face recognition

---
#### ⇒ Algorithm
- An algorithm is the way the model learns patterns from data.
- Common ML Algorithms:
    - Linear Regression
    - Decision Trees
    - Neural Networks
---
- Think:
    - Data = raw material
    - Algorithm = machine processing it
---
- Important Truth:
    - Only Data = Useless
    - Only Algorithm = Useless
    - Both together = ML works.
---
#### ⇒ Pattern
- A pattern is a relationship or structure in data that helps make predictions.
- Examples of patterns:
    - Spam detection
        - Words like “free”, “offer”, “win” → spam
    - Shopping behavior
        - People who buy phone → often buy cover
    - Salary prediction
        - More experience → higher salary
- Key idea
    -Pattern = Hidden rule learned from data


### Traditional Programming and ML

⇒🧾 Traditional Programming
   - Input + Rules → Output
      - You write all rules manually
      - System follows instructions
      - Example:
        - IF marks > 40 → Pass
        - ELSE → Fail
---
⇒ 🤖 Machine Learning
   - Input + Output → Model (learned rules)
   - New Input → Output (prediction)
      - System learns rules automatically
      - No need to hardcode logic


| Aspect              | Traditional Programming | Machine Learning      |
| ------------------- | ----------------------- | --------------------- |
| Rules               | Written by human        | Learned from data     |
| Flexibility         | Low                     | High                  |
| Handling complexity | Hard                    | Easier                |
| Example             | Calculator              | Recommendation system |

- Traditional Programming = You give rules
- Machine Learning = You give data, system finds rules


### ⇒ Why Machine Learning over Traditional Programming?
- Because in many real-world problems, you can’t write all the rules explicitly—but you can learn them from data. That’s where Machine Learning wins over Traditional Programming.

---

#### 1. 🧩 Real-world problems are too complex for rules

##### Traditional Programming (TP):
- Works when logic is clear and fixed
- Fails when patterns are messy or unknown

👉 Example:
- Writing rules for tax calculation → easy
- Writing rules to detect spam emails → nearly impossible

##### Machine Learning (ML):
- Learns hidden patterns automatically
---

#### 2. 🔍 Pattern discovery (not rule writing)

##### In Traditional Programming:
- You define rules

##### In Machine Learning:
- System discovers rules (patterns)
👉 Example (Gmail spam detection):
    - TP: You manually list spam words ❌
    - ML: Learns from millions of emails ✅
---

#### 3. 📈 Improves with more data

##### Traditional Programming:
- Static → doesn’t improve unless you change code

##### Machine Learning:
- Dynamic → improves as more data comes in
👉 Example:
    - Recommendation systems in YouTube get better as you watch more videos

---

#### 4. 🤯 Handles high-dimensional data

##### Traditional Programming struggles with:
- Images
- Audio
- Natural language

##### Machine Learning excels in:
- Face recognition
- Speech recognition
- Chat systems like ChatGPT

---

#### 5. ⚡ Adaptability to change

##### Traditional Programming:
- Needs reprogramming for every change

##### Machine Learning:
- Retrains on new data → adapts automatically

👉 Example:
   - Fraud detection systems constantly evolve

---

#### 6. 🚀 Scalability

##### Traditional Programming:
- More cases → more rules → messy code

##### Machine Learning:
- More data → better model

---

#### ⚖️ When NOT to use Machine Learning

Machine Learning is not always the best choice.
Use Traditional Programming when:
- Rules are simple and fixed
- No data is available
- 100% explainability is required

👉 Example:
   - GST calculation logic → TP is better

---
#### 🧩 Final Comparison

| Situation | Use TP | Use ML |
|----------|--------|--------|
| Clear rules exist | ✅ | ❌ |
| Pattern unknown | ❌ | ✅ |
| Data available | ❌ | ✅ |
| Needs learning/improvement | ❌ | ✅ |

#### 🎯 Final Takeaway

- Use **Traditional Programming** when you know the rules
- Use **Machine Learning** when you only have data, not rules
