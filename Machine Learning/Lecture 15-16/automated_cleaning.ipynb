````python
import pandas as pd
````

````python
df = pd.read_csv("../ML_Datasets/unclean_smartwatch_health_data.csv")
````

````python
df.columns
````

````python
df['Heart Rate (BPM)'].fillna(df['Heart Rate (BPM)'].mean(),inplace=True)
df['Blood Oxygen Level (%)'].fillna(df['Blood Oxygen Level (%)'].mean(),inplace=True)
````

````python
df.drop(columns=['User ID'], inplace=True)
````

#### Forward filling or Backward Filling → Time Series Data

##### F -> 5,6,7,'8',8,9
##### B -> 5,6,7,'7',8,9

- ffill()
- bfill()

```python
df[''].fillna(method='ffill')
df[''].fillna(method='bfill')
```

````python
df['Step Count'].isnull().sum()
````

````python
df['Step Count'] = df['Step Count'].ffill()
````

````python
df['Step Count'].isnull().sum()
````

````python
df[df['Sleep Duration (hours)'] == 'ERROR']
````

````python
df['Sleep Duration (hours)'] = pd.to_numeric(df['Sleep Duration (hours)'], errors='coerce')
````

````python
df['Sleep Duration (hours)'].isnull().sum()
````

````python
df['Sleep Duration (hours)'] = df['Sleep Duration (hours)'].fillna(df['Sleep Duration (hours)'].mean())
````

````python
df['Sleep Duration (hours)'].isnull().sum()
````

Filling Categorical Columns

- Mode ---> Most frequently repeating value in column

````python
df['Activity Level'].mode()
````

````python
df['Activity Level'].isnull().sum()
````

````python
df['Activity Level'].fillna('Seddentary',inplace=True)
````

````python
df['Activity Level'].isnull().sum()
````

### Constant Imputation
- When want to fill missing values with a constant value like 'unknown'

````python
df['Stress Level'].value_counts()
````

````python
df['Stress Level'] = df['Stress Level'].fillna(5, inplace=True)
````
