### Automated Data Cleaning

Automated Data Cleaning refers to the process of using tools, scripts, or algorithms to identify and fix issues in datasets without manual intervention. It helps improve data quality efficiently, especially when working with large datasets.

#### Why Automated Data Cleaning is Important
- Saves time compared to manual cleaning
- Reduces human errors
- Ensures consistency across datasets
- Handles large-scale data efficiently

#### Common Data Cleaning Tasks

1. **Handling Missing Values**
   - Fill missing values using mean, median, or mode
   - Drop rows/columns if too many values are missing

2. **Removing Duplicates**
   - Identify and remove duplicate records to avoid bias

3. **Fixing Data Types**
   - Convert data into correct formats (e.g., string to date, float to integer)

4. **Handling Outliers**
   - Detect extreme values and either remove or cap them

5. **Standardizing Data**
   - Ensure consistency (e.g., "India", "INDIA", "india" → "India")

6. **Encoding Categorical Data**
   - Convert categories into numerical format (Label Encoding, One-Hot Encoding)

#### Example (Using Python - Pandas)

```python
import pandas as pd

# Sample data
data = {
    "Name": ["A", "B", "B", None],
    "Age": [25, None, 30, 22]
}

df = pd.DataFrame(data)

# 1. Handle missing values
df["Age"].fillna(df["Age"].mean(), inplace=True)

# 2. Remove duplicates
df.drop_duplicates(inplace=True)

# 3. Drop rows with missing Name
df.dropna(subset=["Name"], inplace=True)

print(df)

```

### Real Meaning of Automated Data Cleaning

- Automation increases when the system can decide cleaning actions itself.

- Levels of Automation:
    1. Low Automation (Your Example)
        - You define rules
        - System executes
    2. Medium Automation
        - System suggests actions
        - (e.g., detects missing values and recommends mean/median)
    3. High Automation
        - System decides everything automatically
        - (e.g., using ML pipelines)

```python
from sklearn.impute import SimpleImputer

imputer = SimpleImputer(strategy='mean')
df["Age"] = imputer.fit_transform(df[["Age"]])

```
Automation does NOT mean "no code written".

It means:
- Once rules are defined,
- Data cleaning runs automatically
- On any dataset without manual row-by-row intervention

### Patterns of Missing Data

Missing data does not always occur randomly. Understanding the pattern of missing data is important because it affects how we handle and clean the dataset.

There are three main patterns of missing data:

---

#### 1. Missing Completely At Random (MCAR)

Data is missing completely randomly, and the missingness has no relationship with any other variable.

- No pattern in missing data
- Safe to remove or ignore missing values

**Example:**
A survey form where some responses are missing because of accidental skips.

**Impact:**
- Does not bias the dataset
- Simple techniques like dropping rows work fine

---

#### 2. Missing At Random (MAR)

Missing data is related to other observed variables, but not the missing value itself.

- Missingness depends on known data
- Can be handled using statistical methods

**Example:**
Income data is missing more often for younger people.

**Impact:**
- Can introduce bias if ignored
- Use techniques like imputation (mean/median or model-based)

---

#### 3. Missing Not At Random (MNAR)

Missing data is related to the value that is missing itself.

- Missingness has a hidden pattern
- Hardest to handle

**Example:**
People with very high income choose not to disclose their income.

**Impact:**
- Strong bias in data
- Requires advanced methods or domain knowledge

---

### Summary Table

| Type | Description | Bias Risk | Handling |
|------|------------|----------|---------|
| MCAR | Completely random | Low | Drop or simple imputation |
| MAR  | Depends on other variables | Medium | Statistical imputation |
| MNAR | Depends on missing value itself | High | Advanced techniques |

---

### Key Insight

Understanding missing data patterns helps in choosing the right cleaning strategy.
Wrong assumptions can lead to incorrect analysis or poor machine learning models.
