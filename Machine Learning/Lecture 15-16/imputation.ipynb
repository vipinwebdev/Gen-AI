## What is Imputation?

**Imputation** is the process of handling missing values in a dataset by replacing them with meaningful substitutes instead of removing the data.

In real-world datasets, missing values can occur due to errors in data collection, system issues, or incomplete inputs. Imputation helps maintain data completeness and ensures better performance of machine learning models.

---

### 🔑 Why Imputation is Important
- Prevents loss of data (compared to dropping rows/columns)
- Maintains dataset size
- Improves model accuracy
- Enables better statistical analysis

---

### 📌 Common Imputation Techniques

1. **Remove Missing Values**
   Delete rows or columns containing missing data.

2. **Mean / Median Imputation (Numeric Data)**
   Replace missing values with the average (mean) or middle value (median).

3. **Mode Imputation (Categorical Data)**
   Replace missing values with the most frequent value.

4. **Constant / Sentinel Value**
   Replace missing values with a fixed value (e.g., 0, -1, "Unknown").

5. **Forward Fill / Backward Fill (Time Series)**
   Fill missing values using previous or next values.

6. **Simple Imputation (Using Libraries)**
   Use tools like `SimpleImputer` from sklearn for automated filling.

---

### 📊 Example
If a column has values:
`[10, 20, NaN, 30]`

After mean imputation →
`[10, 20, 20, 30]`

---

👉 In short:
**Imputation = Filling missing data with appropriate values instead of deleting it.**

```python
import  pandas as pd
from sklearn.impute import SimpleImputer
```

```python
df = pd.read_csv("../ML_Datasets/unclean_smartwatch_health_data.csv")
```

```python
df.drop(columns=['User ID'], inplace=True)
```

```python
df.columns
```

```python
imputer = SimpleImputer(strategy='median')
```

```python
df.info()
```

```python
df['Heart Rate (BPM)'] = imputer.fit_transform(df[['Heart Rate (BPM)']]).ravel()
```

```python
df.info()
```

```python
imputer1= SimpleImputer(strategy='mean')
```

```python
df['Blood Oxygen Level (%)'] = imputer1.fit_transform(df[['Blood Oxygen Level (%)']]).ravel()
```

```python
imputer2= SimpleImputer(strategy='most_frequent')
```

```python
df['Activity Level'] = imputer2.fit_transform(df[['Activity Level']]).ravel()
```

```python
df.info()
```

```python
imputer3 = SimpleImputer(strategy='constant', fill_value=5)
```

```python
df['Stress Level'] = imputer3.fit_transform(df[['Stress Level']]).ravel()
```

```python
df.info()
```
