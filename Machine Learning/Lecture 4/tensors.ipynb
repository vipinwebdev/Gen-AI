### Tensors in ML

- Tensor is a fundamental unit or data structure in ML or DL or building any AI model.

- A tensor is simply a generalized container for numbers arranged in multiple dimensions.
    - Think of it like an extension of:
        - Scalar → Single number
        - Vector → List of numbers
        - Matrix → Table of numbers
        - Tensor → Multi-dimensional data structure

- A tensor = n-dimensional array of numbers

####  Think of Tensors as “How the World Looks to Machines”

- Humans see:
    - images
    - sounds
    - text

- Machines see:
    - numbers arranged in structured form → tensors

- So tensors are not just data structures—they are how reality is encoded for machine learning.

#### Understanding Step by Step (From Simple to Complex)

1. Scalar (0D Tensor)
    - A single value. → 5
    - No dimension → just one number
    - Example: temperature = 30°C
2. Vector (1D Tensor)
    - A list of values. [1, 2, 3]
    - One dimension
    - Example: features of a house → [area, bedrooms, price]
3. Matrix (2D Tensor)
    - A table (rows × columns)
        - [[1, 2, 3],
          [4, 5, 6]]
        - Two dimensions
        - Example: dataset with rows = samples, columns = features
4. Tensor (3D and above)
    - Now things get interesting 👇
    - `[
  [[1,2], [3,4]],
  [[5,6], [7,8]]
]`
    - 3D tensor
    - Think of it like:
        - Stack of matrices
        - Or multiple images



#### Real-Life Machine Learning Examples

- Example 1: Image Data
    - A color image is stored as a tensor:
    - Height × Width × Channels
    - Example:
        - (224, 224, 3)
    - Meaning:
        - 224 pixels height
        - 224 pixels width
        - 3 channels → RGB
    - 👉 So one image = 3D tensor

- Example 2: Video Data
    - A video is:
        - (Frames, Height, Width, Channels)
        - (30, 224, 224, 3)
    - 👉 Video = 4D tensor






#### 2. Rank (Number of Dimensions)

| Tensor Type | Rank |
| ----------- | ---- |
| Scalar      | 0    |
| Vector      | 1    |
| Matrix      | 2    |
| 3D Tensor   | 3    |

- 👉 Rank = number of axes


#### Why Tensors are Important in ML?
- Because Machine Learning = Mathematical operations on data.
- And tensors allow:
    - Efficient computation (especially on GPUs)
    - Parallel processing
    - Representation of complex data
- 👉 Frameworks like:
    - TensorFlow
    - PyTorch
- use tensors as the core data structure

### Simple Intuition
- Think of tensors like:
    - Boxes inside boxes inside boxes
        - Scalar → one item
        - Vector → one row of items
        - Matrix → table
        - Tensor → multiple tables stacked

### Final Takeaway

A tensor is the fundamental data structure of machine learning, used to represent everything from simple numbers to complex data like images, videos, and text.
