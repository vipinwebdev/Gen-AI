#### Train Test Split

- Train–Test Splitting is the process of dividing your dataset into two parts:
    - Training set → used to train the model
    - Testing set → used to evaluate how well the model performs on unseen data
---
🔹 Why it is needed

If you test the model on the same data it learned from:
👉 it may memorize (overfit) instead of learning patterns

So we keep some data hidden (test set) to check real performance.

---
📊 Common Split
80% Training
20% Testing

(Other ratios like 70–30 also work)

---
🔍 Parameters explained
test_size=0.2 → 20% data for testing
random_state=42 → ensures same split every time (reproducibility)

---
Simple Example
Suppose you have 100 records:

80 → training
20 → testing

Model learns from 80 → tested on 20 → gives accuracy

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=42)
```

What are Evaluation Metrics?

In Regression, we don’t predict categories — we predict numbers.

👉 So we measure:
“How close is the predicted value to the actual value?”

---
1. Mean Squared Error (MSE)



👉 What it does:

Squares the error (difference)
Penalizes large mistakes heavily

✅ Use when large errors are very bad

---
🔹 2. Mean Absolute Error (MAE)



👉 What it does:

Takes absolute difference
Treats all errors equally

✅ Easy to understand
❌ Doesn’t punish big errors extra

---
3. Root Mean Squared Error (RMSE)



👉 What it does:

Square root of MSE
Brings error back to original unit

✅ Most commonly used metric

---
When to use what?
MAE → when you want simple, interpretable error
MSE → when large errors are critical
RMSE → best general-purpose metric
