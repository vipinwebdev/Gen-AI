### sklearn Library

The sklearn library, officially known as Scikit-learn, is one of the most popular machine learning libraries in Python. It provides simple and efficient tools to build, train, and evaluate machine learning models.

🧠 What is Scikit-learn?

Scikit-learn (sklearn) is a high-level ML library built on top of:

- NumPy (for numerical operations)
- SciPy (for advanced math)
- Matplotlib (for visualization)

👉 It is designed to make machine learning easy and accessible, especially for beginners and developers.

⚙️ What can you do with sklearn?

You can perform almost all common ML tasks:

1. 📊 Classification

    - Predict categories (Yes/No, Spam/Not Spam)
    - Example:
        - Email spam detection
        - Disease prediction

    - Algorithms:
        - Logistic Regression
        - Decision Trees
        - Random Forest
---
2. 📈 Regression
    - Predict numerical values
    - Example:
        - House price prediction
        - Salary prediction
---
3. 🔍 Clustering

Group similar data without labels

Example:

Customer segmentation

---
4. 🔽 Dimensionality Reduction

Reduce number of features

Example:

Feature selection
Visualization of high-dimensional data
---

5. 🧹 Data Preprocessing

Prepare data before training

Example:

Scaling (StandardScaler)
Encoding categorical data

---
6. ✅ Model Evaluation

Check model performance

Example:

Accuracy
Precision
Confusion Matrix


```python
from sklearn.linear_model import LinearRegression

# Sample data
X = [[1], [2], [3], [4]]
y = [2, 4, 6, 8]

# Create model
model = LinearRegression()

# Train model
model.fit(X, y)

# Predict
print(model.predict([[5]]))  # Output: ~10
```
